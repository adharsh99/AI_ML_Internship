```python


import pandas as pd

df = pd.read_csv("/expenses.csv")

display(df)
```

```python
# check data information
df.info()
```

```python
# missing values

print("missing values")
print(df.isnull().sum())
```

```python
df.describe
```

```python
# total_expenses
total_expense = df["amount"].sum()

print("Total Expenses:", total_expense)
```

```python
# mean
mean_expenses = df["amount"].mean()
print("mean =",mean_expenses)
```

```python
# highest expenses

highest_expense = df.loc[df["amount"].idxmax()]

print("Highest Expense:")
display(highest_expense)
```

```python
# lowest expense
lowest_expense = df.loc[df["amount"].idxmin()]

print("Lowest Expense:")
display(lowest_expense)
```

```python
# category-wise-expenses
category_expense = df.groupby("category")["amount"].sum()

print("Category-wise Expenses:")
print(category_expense)
```

```python
# category chart
import matplotlib.pyplot as plt

category_expense.plot(kind="bar")

plt.title("Expenses by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
# payment method analysis
payment_expense = df.groupby("payment_method")["amount"].sum()

print("Payment Method-wise Expenses:")
print(payment_expense)
```

```python
# payment method chart
payment_expense.plot(kind="pie", autopct="%1.1f%%")

plt.title("Expenses by Payment Method")
plt.ylabel("")
plt.show()
```

```python
# convert date column
df["date"] = pd.to_datetime(df["date"])

print(df.dtypes)
```

```python
# monthly expense analysis
df["month"] = df["date"].dt.month

monthly_expense = df.groupby("month")["amount"].sum()

print("Monthly Expenses:")
print(monthly_expense)
```

```python
# monthly expense chart
monthly_expense.plot(kind="line", marker="o")

plt.title("Monthly Expenses")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(range(1, 13))
plt.show()
```

```python
# highest spending category
highest_category = category_expense.idxmax()
highest_category_amount = category_expense.max()

print("Highest Spending Category:", highest_category)
print("Amount Spent:", highest_category_amount)
```

```python
# most used payment methods
most_used_payment = df["payment_method"].value_counts().idxmax()
payment_count = df["payment_method"].value_counts().max()

print("Most Used Payment Method:", most_used_payment)
print("Number of Transactions:", payment_count)
```

```python
# FINAL SUMMARY
print("========== EXPENSE ANALYSIS SUMMARY ==========")

print("Total Expense:", df["amount"].sum())
print("Average Expense:", round(df["amount"].mean(), 2))
print("Highest Expense:", df["amount"].max())
print("Lowest Expense:", df["amount"].min())
print("Highest Spending Category:", category_expense.idxmax())
print("Most Used Payment Method:", df["payment_method"].value_counts().idxmax())
```
